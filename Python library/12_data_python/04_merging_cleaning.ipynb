import pandas as pd

df1 = pd.DataFrame({
    "id": [1, 2],
    "name": ["Alice", "Bob"]
})

df2 = pd.DataFrame({
    "id": [1, 2],
    "score": [90, 85]
})

# Merge
merged = pd.merge(df1, df2, on="id")
print(merged)

# Missing values
df = pd.DataFrame({
    "a": [1, None, 3]
})

print(df.isna())

df["a"] = df["a"].fillna(0)
print(df)
