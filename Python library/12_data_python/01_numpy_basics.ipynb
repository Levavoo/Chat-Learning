```python
# Title: NumPy Basics

import numpy as np

# Create arrays
arr = np.array([1, 2, 3])
print(arr)

# Shape
print(arr.shape)

# Arithmetic
print(arr * 2)

# 2D array
matrix = np.array([[1, 2], [3, 4]])
print(matrix)

# Indexing
print(matrix[0, 1])  # 2

# Vectorized operation
print(matrix + 10)
```
