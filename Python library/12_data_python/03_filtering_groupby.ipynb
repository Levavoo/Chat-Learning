import pandas as pd

df = pd.DataFrame({
    "name": ["A", "B", "C", "D"],
    "age": [20, 30, 20, 40],
    "score": [80, 90, 85, 70]
})

# Filtering
print(df[df["age"] > 25])

# Multiple conditions
print(df[(df["age"] > 20) & (df["score"] > 80)])

# Groupby
grouped = df.groupby("age")["score"].mean()
print(grouped)
